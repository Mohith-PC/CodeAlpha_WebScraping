**Import Libraries**

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

**Send Request & Parse HTML**

```python
url = "https://books.toscrape.com"
response = requests.get(url)

soup = BeautifulSoup(response.content, "html.parser")
print("Status Code:", response.status_code)
print("Page Title:", soup.title.text)
```

**Scrape All Pages**

```python
books = []
page = 1

while True:
    url = f"https://books.toscrape.com/catalogue/page-{page}.html"
    response = requests.get(url)

    if response.status_code != 200:
        break  # Stop when no more pages

    soup = BeautifulSoup(response.content, "html.parser")
    articles = soup.find_all("article", class_="product_pod")

    for book in articles:
        title = book.h3.a["title"]
        price = book.find("p", class_="price_color").text.strip()
        rating = book.p["class"][1]  # One, Two, Three, Four, Five
        availability = book.find("p", class_="instock availability").text.strip()

        books.append({
            "Title": title,
            "Price": price,
            "Rating": rating,
            "Availability": availability
        })

    print(f"Scraped page {page}")
    page += 1

print(f"\nTotal books scraped: {len(books)}")
```

**Convert to DataFrame**

```python
df = pd.DataFrame(books)
print(df.shape)
df.head(10)
```

**Basic Analysis**

```python
# Convert price to float
df["Price"] = df["Price"].str.replace("Â£", "").str.replace("£", "").astype(float)

print("Average Price: £", round(df["Price"].mean(), 2))
print("\nRating Distribution:")
print(df["Rating"].value_counts())
print("\nAvailability Count:")
print(df["Availability"].value_counts())
```

**Save to CSV**

```python
 df.to_csv("books_data.csv", index=False)
print("Data saved to books_data.csv")
df.head()
```
